# Comparison Operators

```python
# Equal to
print(5 == 5)    # True
print(5 == '5')  # False (different types)

# Not equal to
print(5 != 3)    # True
print(5 != 5)    # False

# Greater/Less than
print(10 > 5)    # True
print(10 < 5)    # False
print(10 >= 10)  # True
print(10 <= 9)   # False

# String comparison (alphabetical order)
print('apple' < 'banana')  # True (a comes before b)
print('Apple' < 'apple')   # True (ASCII: uppercase < lowercase)
```

# Conditional (Ternary ) Operators

```python
# Traditional if-else
age = 18
if age >= 18:
    status = "Adult"
else:
    status = "Minor"
print(status)  # "Adult"

# Ternary operator (conditional expression)
status = "Adult" if age >= 18 else "Minor"
print(status)  # "Adult"

# More examples
number = 7
result = "Even" if number % 2 == 0 else "Odd"
print(result)  # "Odd"

score = 85
grade = "Pass" if score >= 60 else "Fail"
print(grade)  # "Pass"

# Nested ternary (use sparingly!)
age = 25
category = "Child" if age < 13 else "Teen" if age < 20 else "Adult"
print(category)  # "Adult"
```

```python
# 1. Grade Calculator
score = 85
if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
elif score >= 60:
    grade = "D"
else:
    grade = "F"
print(f"Score: {score}, Grade: {grade}")

# 2. Find primes with break
def is_prime(num):
    if num < 2:
        return False
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            return False
    return True

print(is_prime(17))  # True
print(is_prime(20))  # False

# 3. Clean data with continue
data = [10, -5, 0, 25, -3, 0, 40]
cleaned = []
for num in data:
    if num <= 0:
        continue  # Skip negatives and zeros
    cleaned.append(num)
print(cleaned)  # [10, 25, 40]

# 4. Menu system with break
while True:
    print("\n1. Option A")
    print("2. Option B")
    print("3. Exit")
    choice = input("Choose option: ")
    
    if choice == "1":
        print("You chose A")
    elif choice == "2":
        print("You chose B")
    elif choice == "3":
        print("Goodbye!")
        break
    else:
        print("Invalid choice")
```
