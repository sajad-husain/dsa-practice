# Comparison Operators

```python
# Equal to
print(5 == 5)    # True
print(5 == '5')  # False (different types)

# Not equal to
print(5 != 3)    # True
print(5 != 5)    # False

# Greater/Less than
print(10 > 5)    # True
print(10 < 5)    # False
print(10 >= 10)  # True
print(10 <= 9)   # False

# String comparison (alphabetical order)
print('apple' < 'banana')  # True (a comes before b)
print('Apple' < 'apple')   # True (ASCII: uppercase < lowercase)
```
