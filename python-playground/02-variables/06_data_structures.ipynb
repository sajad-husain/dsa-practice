# Data Structure List, Tuples, Dicts, Sets

# List (mutable, ordered)

```python
fruits = ["apple", "banana", "cheryy"]
print(fruits)
fruits.append("date")
print(fruits)
fruits.insert(1, "kiwi") # insert by index
print(fruits)
fruits.remove("banana") # insert by value
print(fruits)
fruits.pop() # removes and return last item
print(fruits)
fruits.pop(0) # remove and return item at index
print(fruits)
fruits.sort() # sort in place
print(fruits)
fruits.reverse() # reverse in place
print(fruits)
fruits[1:3] # slicing
print(fruits)
fruits[::-1] # reverse via slicing 

print(fruits)
```

# Dictionaries (key value pairs)

```python
person = {"name": "Ali", "age": 25}
person["city"] = "Lahore" # add/update
person.get("age")
person.get("job", "unemployed") # default if key missing( safer than person["job"])
person.pop("age") # remove and return value
"name" in person #True - check key existence
person.keys() # dict keys (['name', 'city'])
person.values() # dict value (['Ali', 'Lahore'])
person.items() # dictionay items ([('name', 'Ali),('city', 'Lahore')])

# merge dictionary
defaults = {"temprature": 0.7, "model": "gpt-4"}
overrides = {"temprature": 0.9}
config = {**defaults, **overrides}
print(config)
```

# Sets (unique, unordered)

```python
tags = {"python", "ai", "ai"}
tags.add("agents")
tags.remove("ai")

a = {1,2,3}
b = {2,3,4}

a | b
a & b
a - b


# Nested structures
chat_history = [
    {"role": "user", "content": "Hello"},
    {"role": "assistant", "content": "Hi there!"}
]

for message in chat_history:
    print(message['content'])
```

```python
inventory = {"apple": 10, "bananas": 5, "oranges": 8}
inventory["grapes"] = 12
inventory["apple"] = 15
inventory.pop("bananas")
  
for key, value in inventory.items():
    print(f"{key}: {value}")
```

```python
list_a = [1,2,3,4,5]
list_b = [4,5,6,7,8]

set_a = set(list_a)
set_b = set(list_b)

print("union", set_a | set_b)
print("intersection", set_a & set_b)
print("difference", set_a - set_b)
```

```python
chat_history = [
    {"role": "user", "content": "Hello"},
    {"role": "assistant", "content": "Hi there!"}
]
for chat in chat_history:
    print(f"{chat['role']} : {chat['content']}")
```
