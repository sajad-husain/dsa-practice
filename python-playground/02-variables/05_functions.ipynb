# Functions, Scope & Modules


```python
def greet(name, greeting="Hlleo"):
    print(greet, name)

greet("Ali")
greet("sajjad", "kese ho")

def add_items(item, cart=None):
    if cart is None:
        cart = []
        cart.append(item)
        return cart

add_items("loads")
```

# Positional vs keyword arguments 

```python
def describe(name, age, marks):
    print(name, age, marks)

describe("sajjad", 27, 88)
```
