# Functions, Scope & Modules


```python
def greet(name, greeting="Hlleo"):
    print(greet, name)

greet("Ali")
greet("sajjad", "kese ho")

def add_items(item, cart=None):
    if cart is None:
        cart = []
        cart.append(item)
        return cart

add_items("loads")
```

# Positional vs keyword arguments 

```python
def describe(name, age, marks):
    print(name, age, marks)

describe("sajjad", 27, 88)
describe("sajjd", age=22, marks=13)
```

# *args and **kwargs 


```python
def sum_all(*args):
    return sum(args)

sum_all(1,2,3,4)

def show_config(**kwargs):
    for key, value in kwargs.items():
        print(key, value)

show_config(model="gpt-4o", temprature= 0.7 )


# combining all arguments (order matters)
def func(a, b, *args, c=10, **kwargs):
    print(a, b, args, c, kwargs)

func(1, 2, 3, 4, c=499, x=100)
```

# Return values

```python
def get_stats(numbers):
    return min(numbers), max(numbers), sum(numbers)/len(numbers)

get_stats([1,2,4,5,6])
```

```python
def calculate_stats(numbers):
    return min(numbers), max(numbers), sum(numbers)/len(numbers)

result = calculate_stats((1,2,3,4,5)) 
for value in result:
    print(value)
    
low, high, average = calculate_stats([1,3,5,6,7])
```

```python
def build_profile(name, **kwargs):
    print(name, kwargs)

build_profile("",Name="Ali", age=25, city="Lahore" )
```
