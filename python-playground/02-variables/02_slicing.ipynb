# Slicing #

```python
new_string = "all is well that end's well"
print(new_string[1:6])
print(new_string[4:-2])
print(new_string[-4:-1])

num_list = [10, 20, 30, 40, 50, 60, 70, 80, 90]
print(num_list[0:90:2])
print(num_list[0:-2:2])
print(num_list[0:-2:3])
print(num_list[::3])
```

# Omitting #

```python
text = "Python Programming"
print(text[:6])
print(text[7:])
print(text[::2]) # every second character
print(text[::-1]) # reverse string
```

# Modifying lists using indexing

```python
color = [ "red", "green", "blue", "skyblue", "orange", "black" ]
color[1] = "pink"
print(color)
color[2:4] = ["black", "black", "black"] # replacing elements
print(color)

color[2:4] = [] #removing elements from the list
print(color)

color[1:1] = [100, 200] # inserting elements in the list 
print(color)
```
