# OOP

```python
class Dog:
    def __init__(self, name, breed): # costructor runs when object is created
        self.name = name             # attribute
        self.breed = breed
    
    def bark(self):                 # method
        return f"{self.name}, says Woof!. It's breed is {self.breed}"
    
my_dog = Dog("Shera", "Poodle")  # creates an instance 

print(my_dog.name)
print(my_dog.bark())
```

# instance attribute vs class attribute 

```python
class Dog:
    species = "Conine" # class attribute shared by ALL instance
    
    def __init__(self, name):
        self.name = name    # instance attribute unique per object 

dog1 = Dog("Rex")
dog2 = Dog("Buddy")

print(dog1.species, dog2.species) # shared
print(dog1.name, dog2.name) # unique 
    
    
```

# Mutable class attribute 

```python
class Dog:
    tricks = [] # DANGER shared across ALL instances!
    
    def add_trick(self, trick):
        self.tricks.append(trick)

d1 = Dog()
d2 = Dog()

d1.add_trick('sit')
print(d2.tricks) # ['sit'] bug d2 got d1's trick too
```

```python
class Dog:
    def __init__(self): # define mutable attributes inside __init__ instead
        self.tricks = [] # fresh list per instance 
    
    def add_tricks(self, trick):
        self.tricks.append(trick)

d1 = Dog()
d2 = Dog()

d1.add_tricks('sit')
d2.add_tricks('stand')

print(d2.tricks)
```

# METHODS

```python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance
    
    def deposit(self, amount):
        self.amount = amount
        self.balance += amount
        return self.balance
    
    def withdraw(self, amount):
        if amount > self.balance:
            raise ValueError("Insufficent funds")
        self.amount = amount
        self.balance -= amount
        return self.balance
    
    def __str__(self): # special method controls what print() shows
        return f"{self.owner}'s account: ${self.balance}"
    
acc = BankAccount("Ali", 100)

acc.deposit(50)
print(acc)
acc.withdraw(20)
print(acc)
```

# Inheritance

```python
class Animal:
    def __init__(self, name):
        self.name = name
    
    def speak(self):
        return "some rendom sound"

class Dog(Animal):
    def speak(self):
        return f"{self.name} says Woof!"

class Cat(Animal):
    def speak(self):
        return f"{self.name} says Meow!"

all_animals = [ Dog("Rex"), Cat("Pishi")]
for a in all_animals:
    print(a.speak())
```

```python
class Animal:
    def __init__(self, name):
        self.name = name
        self.alive = True

class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name) # calls animals init first 
        self.breed = breed # than adds dog specific attributes

d = Dog("Rex", "Poodle")

print(d.name, d.alive, d.breed)
```

# Encapsulation

```python
class Account:
    def __init__(self, balance):
        self._balance = balance     #single underscore = "internal use"
        self.__secret = "hidden"    #double underscore = "name mangled, harder to access"
    
    def get_balance(self):
        return self._balance
```

# class methods vs static methods

```python
class MathHelper:
    @staticmethod
    def add(a,b):           # does'nt need self or any instance data
        return a + b
    
    @classmethod
    def from_string(cls, s):
        a, b = map(int, s.split(","))  # receives the class itself, not an instance
        return cls(a,b)
print(MathHelper.add(2, 3)) # called without creating an instance
```

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def greet(self):
        return f"Hi, I'm {self.name} and I'm {self.age} years old"
    
    def __str__(self):
        return f"Hi, I'm {self.name} and I'm {self.age} years old"

class Student(Person):
    def __init__(self, school):
        self.school = school
        super().__init__()
    
    def greet(self):
        return f"Hi, I'm {self.name} and I'm {self.age} years old. My school name is {self.school}"
```
