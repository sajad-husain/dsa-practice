# OOP

```python
class Dog:
    def __init__(self, name, breed): # costructor runs when object is created
        self.name = name             # attribute
        self.breed = breed
    
    def bark(self):                 # method
        return f"{self.name}, says Woof!. It's breed is {self.breed}"
    
my_dog = Dog("Shera", "Poodle")  # creates an instance 

print(my_dog.name)
print(my_dog.bark())
```

# instance attribute vs class attribute 

```python
class Dog:
    species = "Conine" # class attribute shared by ALL instance
    
    def __init__(self, name):
        self.name = name    # instance attribute unique per object 

dog1 = Dog("Rex")
dog2 = Dog("Buddy")

print(dog1.species, dog2.species) # shared
print(dog1.name, dog2.name) # unique 
    
    
```
