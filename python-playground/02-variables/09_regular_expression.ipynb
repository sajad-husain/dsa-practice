# Regular Expressions (Regex)

```python
import re

text = "My phone number is 123-456-7890"

match = re.search(r"\d{3}-\d{3}-\d{4}", text)
if match:
    print(match.group())
```

```python
# search() find first match anywhere in the string
import re
re.search(r"\d+", "abc 123 def")

# match() only checks START of the string
matching1 = re.match(r"\d+", "123 abc")
matching2 = re.match(r"\d+", "abc 123")
print("match1 =>", matching1,"match 2",matching2)

# findall() returns all matches as a list
re.findall(r"\d+", "a1 b22 c333")

# sub() substitute/replaces matches
re.sub(r"\d+", "#", "a1 b22 c333")

# split() split string by pattern
re.split(r"\s+", "hello world fooo is       well")
```

```python
# groups extracting parts of the match
import re
text = "Emails: ali@example.com jd@example.com sh@example.com"
match = re.search(r"(\w+)@(\w+)\.(\w+)", text)
finding_match = re.findall(r"(\w+)@(\w+)\.(\w+)", text)
print(match.group(0))
```
