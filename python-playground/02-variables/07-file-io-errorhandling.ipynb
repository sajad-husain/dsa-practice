# Writing a file 

```python
with open("notes.txt", "w") as f:
    f.write("Hello, file!\n")
    f.write("second line.\n")
```

# Reading a file 

```python
with open("notes.txt", "r") as f:
    read_content = f.read() # entire file as one string
    print(read_content)

with open("notes.txt", "r") as f:
    read_lines = f.readlines() # list of lines
    print(read_lines)
    
with open("notes.txt","r") as f:
    for line in f: # memory efficent, reads line by line
        print(line.strip()) # .strip() removes trailing \n
```

# Apending

```python
with open("notes.txt", "a") as f:
    f.write("Append this line \n")
```

# Error handling try/except

```python
try:
    result = 10/0
except ZeroDivisionError:
    print("Can't divide by zero!")
```

```python
# BAD PRACTICE
try:
    risky_operations()
except:
    pass

# GOO PRACTICE
try:
    risky_operations()
except (ValueError, KeyError) as e:
    print(f"Something specific went wrong: {e}")
```

```python
try:
    f = open("data.txt", "r")
except FileNotFoundError:
    print("File not found!")
else:
    print("File opened successfully") # runs only if no exceptions occured
    f.close()
finally:
    print("this always runs") # runs no matter what 
    
'''
ValueError => wrong value type for an operation, e.g. int("abc")
TypeError => wrong data type used, e.g. "5" + 5
KeyError => dict key doesn't exist
IndexError => list index out of range
FileNotFoundError => file doesn't exist
ZeroDivisionError => dividing by zero
AttributeError => calling a method that doesn't exist on an object
'''
```

```python
def set_age(age):
    if age < 0:
        raise ValueError("Age can't be negative")
    return age

try:
    set_age(-4)
except ValueError as e:
    print(f"Error: {e}")
```

```python
class InsufficentFundsError(Exception):
    pass

def withdraw(balance, amount):
    if amount > balance:
        raise InsufficentFundsError("Note enough balance!")
    return balance - amount
```
