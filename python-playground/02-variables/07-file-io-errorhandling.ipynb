# Writing a file 

```python
with open("notes.txt", "w") as f:
    f.write("Hello, file!\n")
    f.write("second line.\n")
```

# Reading a file 

```python
with open("notes.txt", "r") as f:
    read_content = f.read() # entire file as one string
    print(read_content)

with open("notes.txt", "r") as f:
    read_lines = f.readlines() # list of lines
    print(read_lines)
    
with open("notes.txt","r") as f:
    for line in f: # memory efficent, reads line by line
        print(line.strip()) # .strip() removes trailing \n
```

# Apending

```python
with open("notes.txt", "a") as f:
    f.write("Append this line \n")
```
